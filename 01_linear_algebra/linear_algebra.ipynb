## سؤال‌های مصاحبه‌ای
1. تفاوت `A * B` و `A @ B`؟
2. اگر دو بردار عمود باشند، ضرب داخلی چقدر است؟
3. چرا `v.shape` برای بردار `(3,)` است و نه `(3, 1)`؟

```python
import numpy as np

v = np.array([1, 2, 3])
print(f"vector : {v}")
print(f"shape : {v.shape}")

M = np.array([[1, 2, 3],
            [4, 5, 6]])
print(f"matris : {M}")
print(f"shape : {M.shape}")
```

```python
# a = np.array([1, 2, 3])
# b = np.array([4, 5, 6])

a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])

dot = np.dot(a, b)
print(f"Inner product : {dot}")
print(f"with @ ----> a @ b : {a@b}")
```

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])


print(f"Element-wise multiplication ---> A * B : {A * B} ")
```

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

print(f"A : \n{A}")
print(f"A.T : \n{A.T}")

v = np.array([3, 4])

print(f"norm vector : {np.linalg.norm(v)}")
##sqrt(3²+4²) = 5
```

```python
p1 = np.array([1, 2, 3])
p2 = np.array([4, 6, 8])

distance = np.linalg.norm(p1-p2)

print(f"Euclidean : {distance}")
```
