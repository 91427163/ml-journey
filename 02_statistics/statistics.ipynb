```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 1000])

print(f'average : {np.mean(data)}')     
print(f'median : {np.median(data)}')
```

```python
data = np.array([10, 20, 30, 40, 50])

print(f'var : {np.var(data)}')     
print(f'std : {np.std(data)}')     
```

```python
rng = np.random.default_rng(42)
samples = rng.normal(loc=0, scale=1, size=10000)

print(f'average : {np.mean(samples)}')     
print(f'std : {np.std(samples)}')     
print(f"68% between -1 < n < 1 : {np.mean((samples > -1) & (samples < 1))}")
print(f"95% between -2 < n < 2 : {np.mean((samples > -2) & (samples < 2))}")
```

```python
rng = np.random.default_rng(42)
x = rng.normal(0, 1, 1000)
y = 2 * x + rng.normal(0, 0.5, 1000)

print(f"همبستگی : {np.corrcoef(x, y)[0, 1]}")
```

### قضیه بیز (مثال تشخیص بیماری)

```python
p_disease = 0.01
p_pos_given_disease = 0.99
p_pos_given_healthy = 0.05

p_pos = p_pos_given_disease * p_disease + p_pos_given_healthy * (1 - p_disease)

p_disease_given_pos = (p_pos_given_disease * p_disease) / p_pos

print(f"احتمال بیماری با تست مثبت: {p_disease_given_pos:.2%}")
```
