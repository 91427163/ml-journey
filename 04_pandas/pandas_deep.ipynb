```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Reza', 'Mina', 'Hassan', 'Neda'],
    'city': ['Tehran', 'Tehran', 'Shiraz', 'Shiraz', 'Tehran', 'Mashhad'],
    'age': [25, 30, 35, np.nan, 28, 42],
    'salary': [5000, 7000, 6000, 8000, np.nan, 9000]
})

df
```

```python
print(df.info())
print("*"*10)
print(df.describe())
```

```python
print("میانگین حقوق هر شهر:")
print(df.groupby('city')['salary'].mean())

# چند تجمیع همزمان
print("\nچند تجمیع:")
print(df.groupby('city')['salary'].agg(['mean', 'count', 'sum']))
```

```python
print("تعداد NaN در هر ستون:")
print(df.isna().sum())


# گزینه ۱: حذف ردیف‌های NaN
print("\nحذف NaN:")
print(df.dropna())

# گزینه ۲: پر کردن با میانگین
print("\nپر کردن با میانگین:")
df_filled = df.copy()

df_filled['age'] = df_filled['age'].fillna(df_filled['age'].mean())
df_filled['salary'] = df_filled['salary'].fillna(df_filled['salary'].median())
df_filled
```

```python
df2 = pd.DataFrame({
    'city': ['Tehran', 'Shiraz', 'Mashhad', 'Tabriz'],
    'population': [9_000_000, 1_800_000, 3_000_000, 1_600_000]
})

merged = df.merge(df2, on='city', how='left')
merged
```
