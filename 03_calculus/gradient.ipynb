```python
def f(x):
    return x ** 2

def derivative(f, x, h = 1e-5):
    return (f(x+h) - f(x-h))/(2 * h)

for x in [-2, 0, 3]:
    print(f"f({x}) : {f(x)}, f'({x}) : {derivative(f, x):.2f}")
```

```python
def f2(x, y):
    return x**2 + 3*y

def partial_x(f, x, y, h=1e-5):
    return (f(x + h, y) - f(x - h, y)) / (2 * h)
def partial_y(f, x, y, h=1e-5):
    return (f(x, y + h) - f(x, y - h)) / (2 * h)

print("∂f/∂x در (2, 3):", partial_x(f2, 2, 3))
print("∂f/∂y در (2, 3):", partial_y(f2, 2, 3))
```

```python
def f(x):
    return (x - 3)**2  # کمینه در x=3

def grad_f(x):
    return 2 * (x - 3)

x = 0.0           # نقطه شروع
lr = 0.1          # نرخ یادگیری
for i in range(20):
    x = x - lr * grad_f(x)
    if i % 5 == 0:
        print(f"گام {i}: x = {x:.3f}, f(x) = {f(x):.3f}")

print(f"نتیجه نهایی: x = {x:.3f}")
```

```python
def run_gd(lr, steps=20):
    x = 0.0
    for _ in range(steps):
        x = x - lr * grad_f(x)
    return x

for lr in [0.01, 0.1, 0.5, 1.01]:
    result = run_gd(lr)
    print(f"lr={lr}: x نهایی = {result:.3f}")
```
